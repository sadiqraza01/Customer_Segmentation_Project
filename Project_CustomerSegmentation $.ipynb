```python
# Select only numeric columns
numeric_cols = df.select_dtypes(include='number').columns

# Store outlier info
outlier_summary = {}

# Loop through each numeric column
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
    outlier_count = outliers.shape[0]
    outlier_summary[col] = outlier_count

# Show outlier count per column
print("Outlier count in each column:")
for col, count in outlier_summary.items():
    print(f"{col}: {count} outliers")
```

```python
# Make a copy for cleaning
df_clean = df.copy()

# Loop through each numeric column and filter out outliers
for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    df_clean = df_clean[(df_clean[col] >= lower_limit) & (df_clean[col] <= upper_limit)]

# Result
print(f"Original shape: {df.shape}")
print(f"Cleaned shape: {df_clean.shape}")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
#  Feature Engineering

 # Age from Year_Birth
df_clean['Age'] = 2025 - df_clean['Year_Birth'] 

#Total spends column
spend_cols = ['MntWines', 'MntFruits', 'MntMeatProducts',
              'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']

df_clean['Total_Spent'] = df_clean[spend_cols].sum(axis=1)  # Total money spent
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='Income', y='Total_Spent', color='royalblue')

plt.title("Scatter Plot: Income vs Total_Spent")
plt.xlabel("Income")
plt.ylabel("Total Spent")
plt.grid(True)
plt.show()
```

```python
# One-hot encode categorical columns
df_encoded = pd.get_dummies(df_clean[['Education', 'Marital_Status']], drop_first=True)
```

```python
# Select features

numeric_features = ['Income', 'Age', 'Recency', 'Total_Spent',
                    'NumWebPurchases', 'NumStorePurchases', 'NumCatalogPurchases',
                    'Kidhome', 'Teenhome']
```

```python
# Combine numerical + encoded features
X = pd.concat([df_clean[numeric_features], df_encoded], axis=1)
```

```python
# Drop missing values
X = X.dropna()
```

```python
X
```

```python
# Scale the features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# Using Elbow method to find optimal K

import warnings
warnings.filterwarnings('ignore')

inertia = [] 
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
# plotting elbow curve

plt.figure(figsize=(8, 4))
sns.lineplot(x=range(1, 11), y=inertia, marker='o')
plt.title("Elbow Method For Optimal k")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
```

```python
# fitting final KMeans model with k=3 is best

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
# Adding cluster labels to the original data
df_clean['Cluster'] = kmeans_labels
```

```python
# Visualize clusters (example with Income & Total_Spent)
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_clean['Income'], y=df_clean['Total_Spent'], hue=df_clean['Cluster'], palette="Set2")
plt.title("Customer Segments (Income vs Spending)")
plt.xlabel("Income")
plt.ylabel("Total Spending")
plt.grid(True)
plt.show()
```

MODEL 2: Gaussian Mixture Model (GMM) 

```python
from sklearn.mixture import GaussianMixture

# Fiting GMM Model
gmm = GaussianMixture(n_components=3, random_state=42)  

gmm_labels = gmm.fit_predict(X_scaled)
```

```python
# Adding cluster labels to the original data
df_clean['GMM_Cluster'] = gmm_labels
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df_clean['GMM_Cluster'], x=df_clean['Income'], y=df_clean['Total_Spent'], hue=df_clean['GMM_Cluster'], palette='Set2', s=60)
plt.title("GMM Clusters Based on Income vs Spending")
plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

#MODEL 3

# Create the model
agglo = AgglomerativeClustering(n_clusters=3)  
agglo_labels = agglo.fit_predict(X_scaled)

# Add cluster labels to df_clean
df_clean['Agglo_Cluster'] = agglo_labels
```

```python
sns.scatterplot(x=df_clean['Income'],y=df_clean['Total_Spent'],hue=df_clean['Agglo_Cluster'],palette="Set2")
plt.title("Agglomerative Clustering")
plt.xlabel("Income")
plt.ylabel("Total Spending")
plt.legend(title='Cluster')
plt.show()
```

```python
from sklearn.cluster import DBSCAN

# Create DBSCAN model
dbscan = DBSCAN(eps=1.2, min_samples=5) #A group of at least 5 nearby points is considered a cluster

# Fit and predict
dbscan_labels = dbscan.fit_predict(X_scaled)
df_clean['DBSCAN_Cluster']=dbscan_labels
```

```python
sns.scatterplot(x=df_clean['Income'],y=df_clean['Total_Spent'],hue=df_clean['DBSCAN_Cluster'],palette="Set2")
plt.title("DBSCAN Clustering")
plt.xlabel("Income")
plt.ylabel("Total Spending")
plt.legend(title='Cluster')
plt.show()
```

```python

from sklearn.metrics import silhouette_score


score = silhouette_score(X_scaled, kmeans_labels)
print(f"Silhouette Score for KMeans model: {score:.3f}")



score = silhouette_score(X_scaled, gmm_labels)
print(f"Silhouette Score for Gaussian Mixture Model: {score:.3f}")


score = silhouette_score(X_scaled, agglo_labels)
print(f"Silhouette Score for Agglomerative Clustering: {score:.3f}")


score = silhouette_score(X_scaled, dbscan_labels)
print(f"Silhouette Score for DBSCAN Clustering: {score:.3f}")
```

| Model                | Silhouette Score | Meaning                                                  |
| -------------------- | ---------------- | -------------------------------------------------------- |
| **KMeans**           | 0.116            | 😐 Not very good, clusters are slightly overlapping      |
| **Gaussian Mixture** | 0.099            | 😐 Similar to KMeans, even more overlap between clusters |
| **Agglomerative**    | 0.183            | ✅ Best result so far, clusters are better separated      |
| **DBSCAN**           | -0.194           | ❌ Bad clustering, many points are likely mislabeled      |
